```python
# Raiz quadrada para calcular a distancia
import math

import pandas as pd

# Classificação com KNN
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Euclidiana
# Cliente A e Cliente B : renda e historico de pagament
p1 = (2.0, 3.0) # Clite A renda baixa, historico ruim
p2 = (8.0, 7.0) # CLiente B renda alta, historico bom

# Diferença em cada dimensão (cateto do triangulo retangulo)
dx = p1[0] - p2[0]
dy = p1[1] - p2[1]

# Distancia = Hipotenusa (Pitagoras)
distancia = math.sqrt(dx**2 + dy**2)
distancia_manhattan = abs(dx) + abs(dy)

print(f"Euclidiana: {distancia:.2f}")
print(f"Manhattan: {distancia_manhattan:.2f}")
print()
print("Euclidiana: distancia em linha reta.")
print("Manhattan: soma dos passos em grade (horizontal + vertical).")
```
